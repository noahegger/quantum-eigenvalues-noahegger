```python
file_name = 'harmonic_oscillator_wf.dat'

# loadtxt reads all the data in a file and stores them in an array (assuming it's all floats)
# we use skiprows=1 to avoid the header
all_data = np.loadtxt(file_name,skiprows=1)

# We transpose the array in order to be able to extract each column individually
all_data = all_data.transpose()
x = all_data[0]
ground_harmonic = all_data[1]
first_excited_harmonic = all_data[2]
second_excited_harmonic = all_data[3]
```

```python
file_name = 'woods_saxon_wf.dat'

# loadtxt reads all the data in a file and stores them in an array (assuming it's all floats)
# we use skiprows=1 to avoid the header
all_data = np.loadtxt(file_name,skiprows=1)

# We transpose the array in order to be able to extract each column individually
all_data = all_data.transpose()
x = all_data[0]
ground_saxon = all_data[1]
first_excited_saxon = all_data[2]
second_excited_saxon = all_data[3]
```

```python
plt.plot(x, ground, label='Inf. Well')
plt.plot(x, ground_harmonic, label='Harmonic')
plt.plot(x, ground_saxon, label='Saxon')
# Set appropriate labels with the correct units.
plt.title('Ground State')
plt.xlabel('Location between -L and L (fm) ')
plt.ylabel('Probability Density (1/m)')
plt.legend(loc='upper right')
plt.show()
```

```python

plt.plot(x, first_excited, label='Inf. Well')
plt.plot(x, first_excited_harmonic, label='Harmonic')
plt.plot(x, first_excited_saxon, label='Saxon')

# Set appropriate labels with the correct units.
plt.title('First Excited State')
plt.xlabel('Location between -L and L (fm) ')
plt.ylabel('Probability Density (1/m)')
# legend displays the labels given in each plot
plt.legend(loc='upper right')
plt.show()
```

```python

plt.plot(x, second_excited, label='Inf. Well')
plt.plot(x, second_excited_harmonic, label='Harmonic')
plt.plot(x, second_excited_saxon, label='Saxon')
# Set appropriate labels with the correct units.
plt.title('Second Excited State')
plt.xlabel('Location between -L and L (fm)')
plt.ylabel('Probability Density (1/m)')
# legend displays the labels given in each plot
plt.legend(loc='upper right')
plt.show()
```

```python
file_name = 'woods_saxon_ener.dat'

# loadtxt reads all the data in a file and stores them in an array (assuming it's all floats)
# we use skiprows=1 to avoid the header
all_data = np.loadtxt(file_name,skiprows=1)

# We transpose the array in order to be able to extract each column individually
all_data = all_data.transpose()

# You should know what to do
r = all_data[0]
ground_r = all_data[1]
first_excited_r = all_data[2]
second_excited_r = all_data[3]
```

```python
plt.plot(r, ground_r, label='Ground')
plt.plot(r, first_excited_r, label='First Excited')
plt.plot(r, second_excited_r, label='Second Excited')
# Set appropriate labels with the correct units.
plt.title('Energy v.s. Radius')
plt.xlabel('Radius (fm) ')
plt.ylabel('Energy (MeV)')
plt.legend(loc='upper right')
plt.show()
```
